```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Parameters
T = 5
r = 0.03
X0 = 10
gamma1 = 0.5

gamma2_values = [0.005, 0.01]

G = np.array([1.0, -0.5])

mu = np.array([0.12, 0.22])
theta = mu - r * np.ones(2)

# Volatility parameters
sigma1 = 0.16
sigma2 = 0.48
rho = 0.09

sigma = np.array([
    [sigma1, 0.0],
    [sigma2 * rho, sigma2 * np.sqrt(1 - rho**2)]
])

Sigma = sigma @ sigma.T
Sigma_inv = np.linalg.inv(Sigma)

# Local mean-variance component
pi_MV = (1 / (gamma1 * X0)) * np.exp(-r * T) * Sigma_inv @ theta

rows=[]

# Brown investor
for i in range(2):
    rows.append({
        "Investor": "Brown",
        "Asset": f"Asset {i+1}",
        "Local MV": pi_MV[i],
        "ESG": np.nan,
        "Combined": pi_MV[i]
    })

# Green investors
for gamma2 in gamma2_values:
    pi_ESG = (gamma2 / (gamma1**2 * X0)) * np.exp(-r * T) * Sigma_inv @ G
    pi_green = pi_MV + pi_ESG

    for i in range(2):
        rows.append({
            "Investor": f"Green, gamma2={gamma2}",
            "Asset": f"Asset {i+1}",
            "Local MV": pi_MV[i],
            "ESG": pi_ESG[i],
            "Combined": pi_green[i]
        })

table = pd.DataFrame(rows)
print(table.round(4))
```

```python
import numpy as np

def simulate_wealth(X0,mu,sigma,gamma1,gamma2,Sigma,G,r,theta, T,Ndt,Nsims=100):
    dt=T/Ndt
    time = np.linspace(0,T,Ndt+1)

    X_b=np.zeros((Ndt+1,Nsims))
    X_g=np.zeros((Ndt+1,Nsims))

    X_b[0,:]=X0
    X_g[0,:]=X0

    Sigma_inv=np.linalg.inv(Sigma)

    for idx in range(Ndt):
        t=time[idx]

        #Optimal amount for brown investor
        u_b=(1/gamma1)*np.exp(-r*(T-t))*Sigma_inv@theta

        #Optimal amount for green investor
        u_g=((1/gamma1)*np.exp(-r*(T-t))*Sigma_inv@theta+(gamma2/gamma1**2)*np.exp(-r*(T-t))*Sigma_inv@G)

        Z=np.random.randn(2,Nsims)
        dW=np.sqrt(dt)*Z

        #Euler discretization
        X_b[idx+1,:]=(X_b[idx,:]+(r*X_b[idx,:]+u_b.T@theta)*dt+(u_b.T@sigma@dW))

        X_g[idx+1,:]=(X_g[idx,:]+(r*X_g[idx,:]+u_g.T@theta)*dt+(u_g.T@sigma@dW))

    return time, X_b,X_g


# Parameters
T = 10
Ndt=500
Nsims=10000
r = 0.03
X0 = 10
gamma1 = 0.5
gamma2=0.01

G = np.array([1.0, -0.5])

mu = np.array([0.12, 0.22])
theta = mu - r * np.ones(2)

# Volatility parameters
sigma1 = 0.16
sigma2 = 0.48
rho = 0.09

sigma = np.array([
    [sigma1, 0.0],
    [sigma2 * rho, sigma2 * np.sqrt(1 - rho**2)]
])

Sigma = sigma @ sigma.T
Sigma_inv = np.linalg.inv(Sigma)

time, X_b, X_g=simulate_wealth(
    X0=X0,
    r=r,
    T=T,
    mu=mu,
    theta=theta,
    sigma=sigma,
    Sigma=Sigma,
    G=G,
    gamma1=gamma1,
    gamma2=gamma2,
    Ndt=Ndt,
    Nsims=Nsims
)

plt.figure(figsize=(10, 6))
plt.plot(time, X_b.mean(axis=1), label="Brown investor")
plt.plot(time, X_g.mean(axis=1), label="Green investor")
plt.xlabel("Time")
plt.ylabel("Mean wealth")
plt.title("Mean simulated wealth under optimal strategies")
plt.legend()
plt.grid(True)
plt.show()

    
```

We now want to examine how changes in the different parameters affect the wealth for both brown and green investors.
We will try to vary $\gamma_2$, G and $\mu$.

```python
np.random.seed(123)

#Vary gamma2
gamma2_values=[0, 0.005, 0.01, 0.02]
path=0

plt.figure(figsize=(10, 6))

for gamma2_v in gamma2_values:
    np.random.seed(123)
    
    time, X_b, X_g=simulate_wealth(
    X0=X0,
    r=r,
    T=T,
    mu=mu,
    theta=theta,
    sigma=sigma,
    Sigma=Sigma,
    G=G,
    gamma1=gamma1,
    gamma2=gamma2_v,
    Ndt=Ndt,
    Nsims=Nsims,
)

    if gamma2_v==0:
        plt.plot(time, X_g[:,path],label="Brown investor, $\gamma_2=0$", color="brown")
    else:
        plt.plot(
            time,
            X_g[:, path],
            label = fr"Green investor, $\gamma_2={gamma2_v}$"
        )

plt.xlabel("Time")
plt.ylabel("Wealth")
plt.title("Simulated wealth paths for brown and green investors")
plt.grid(True)
plt.legend()
plt.show()
```

```python
np.random.seed(123)

#Vary gamma2
gamma2_values=[0, 0.005, 0.01, 0.02]
path=0

plt.figure(figsize=(10, 6))

for gamma2_v in gamma2_values:
    np.random.seed(123)
    
    time, X_b, X_g=simulate_wealth(
    X0=X0,
    r=r,
    T=T,
    mu=mu,
    theta=theta,
    sigma=sigma,
    Sigma=Sigma,
    G=G,
    gamma1=gamma1,
    gamma2=gamma2_v,
    Ndt=Ndt,
    Nsims=Nsims,
)
    mean_difference = X_g.mean(axis=1) - X_b.mean(axis=1)

    plt.plot(time, mean_difference, label=fr"$\gamma_2={gamma2_v}$")


plt.axhline(0, linestyle="--")
plt.xlabel("Time")
plt.ylabel("Mean wealth difference")
plt.title(r"Effect of varying $\gamma_2$: $\mathbb{E}[X_t^{g,*}]-\mathbb{E}[X_t^{b,*}]$")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#We now want to see what happens if we vary G to see the effect of ratings.
gamma2=0.01

G_values=[
    np.array((1, -0.5)),
    np.array([1, -1]),
    np.array([0.5, -0.5]),
    np.array([0.5, -1])
]

G_labels = [
    r"$G=(1,-0.5)$",
    r"$G=(1,-1)$",
    r"$G=(0.5,-0.5)$",
    r"$G=(0.5,-1)$"
]

n_paths=1
plt.figure(figsize=(10, 6))

#Brown investor is not affected by G
np.random.seed(123)
time, X_b, _=simulate_wealth(
    X0=X0,
    r=r,
    T=T,
    mu=mu,
    theta=theta,
    sigma=sigma,
    Sigma=Sigma,
    G=G_values[0],
    gamma1=gamma1,
    gamma2=gamma2,
    Ndt=Ndt,
    Nsims=Nsims,
)

for path in range(n_paths):
    plt.plot(
        time,X_b[:,path],color='brown', alpha=0.4,label="Brown investor" if path == 0 else None
    )

#Green investor with fixed gamma2 and varying G
for G_v, G_label in zip(G_values, G_labels):
    np.random.seed(123)

    time, _, X_g = simulate_wealth(
        X0=X0,
        r=r,
        T=T,
        mu=mu,
        theta=theta,
        sigma=sigma,
        Sigma=Sigma,
        G=G_v,
        gamma1=gamma1,
        gamma2=gamma2,
        Ndt=Ndt,
        Nsims=Nsims,
    )
    for path in range(n_paths):
        plt.plot(time, X_g[:,path], alpha =0.5, label=G_label if path == 0 else None)

plt.xlabel("Time")
plt.ylabel("Wealth")
plt.title(fr"Simulated wealth paths for varying G")
plt.grid(True)
plt.legend()
plt.show()
```

```python
#Now we try to vary $\mu$. Changing mu affects both the brown and green investor

mu_values = [
    np.array([0.12, 0.22]),  # low ESG asset has higher return
    np.array([0.12, 0.12]),  # equal expected returns
    np.array([0.22, 0.12])   # high ESG asset has higher return
]

mu_labels = [
    r"$\mu=(0.12,0.22)$",
    r"$\mu=(0.12,0.12)$",
    r"$\mu=(0.22,0.12)$"
]

n_paths=1
gamma2=0.01

plt.figure(figsize=(10,6))

for mu_v,mu_label in zip(mu_values, mu_labels):
    theta_v=mu_v-r*np.ones(2)

    np.random.seed(123)

    time, X_b,X_g = simulate_wealth(
        X0=X0,
        r=r,
        T=T,
        mu=mu_v,
        theta=theta_v,
        sigma=sigma,
        Sigma=Sigma,
        G=G,
        gamma1=gamma1,
        gamma2=gamma2,
        Ndt=Ndt,
        Nsims=Nsims,
    )
    for path in range(n_paths):
        plt.plot(time, X_b[:,path],linestyle="--", alpha=0.4, label="Brown, "+mu_label if path == 0 else None)

        plt.plot(
            time, X_g[:,path],alpha=0.5, label="Green, "+ mu_label if path == 0 else None)

plt.xlabel("Time")
plt.ylabel("Wealth")
plt.title(fr"Simulated wealth paths for varying $\mu$")
plt.grid(True)
plt.legend()
plt.show()
```

```python
gamma2_values = [0.0, 0.005, 0.01, 0.02]

results = []

for gamma2_v in gamma2_values:
    np.random.seed(123)

    time, X_b, X_g = simulate_wealth(
        X0=X0,
        r=r,
        T=T,
        mu=mu,
        theta=theta,
        sigma=sigma,
        Sigma=Sigma,
        G=G,
        gamma1=gamma1,
        gamma2=gamma2_v,
        Ndt=Ndt,
        Nsims=Nsims,
    )

    terminal_b = X_b[-1, :]
    terminal_g = X_g[-1, :]
    diff = terminal_g - terminal_b

    results.append({
        "gamma2": gamma2_v,
        "mean_XT_brown": terminal_b.mean(),
        "mean_XT_green": terminal_g.mean(),
        "mean_difference": diff.mean(),
        "std_difference": diff.std(),
        "prob_green_higher": np.mean(diff > 0)
    })


results_table=pd.DataFrame(results)
results_table = results_table.rename(columns={
    "gamma2": "$\gamma_2$",
    "mean_XT_brown": "Mean terminal wealth brown",
    "mean_XT_green": "Mean terminal wealth green",
    "mean_difference": "Mean difference",
    "std_difference": "Std. difference",
    "prob_green_higher": "P(Green > Brown)"
})

small_table = results_table[
    [
        r"$\gamma_2$",
        "Mean terminal wealth brown",
        "Mean terminal wealth green",
        "Mean difference"
    ]
]

print(small_table)
```
